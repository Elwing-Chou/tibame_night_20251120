```python
# 五子棋
# 1. 先把棋盤畫出來
# 2. 落子(落到最相鄰的位置)
# 3. 判斷落子是否五子
import pygame as pg

#pygame初始化
pg.init()
#設定視窗
width, height = 640, 640
# 棋盤間隔
x_inter, y_inter = width / 16, height / 16

# 產生視窗
screen = pg.display.set_mode([width, height])
# 設定遊戲標題
pg.display.set_caption("五子棋")
# 建立畫布bg
bg = pg.Surface(screen.get_size())
# 把畫布填滿某個顏色
bg.fill([199, 167, 82])


board_line_color = [100, 100, 100]
i = 0
while i < 15:
    pg.draw.line(bg,
                 board_line_color,
                 [x_inter, y_inter*i+y_inter],
                 [width-x_inter, y_inter*i+y_inter],
                 2)
    i = i + 1

i = 0
while i < 15:
    pg.draw.line(bg,
                 board_line_color,
                 [x_inter*i+x_inter, y_inter],
                 [x_inter*i+x_inter, height-y_inter],
                 2)
    i = i + 1

# 把某個圖層貼到上一層
screen.blit(bg, [0, 0])
# 對畫面進行更新(才會真的秀出來)
pg.display.update()

game_round = 0

running = True
while running:
    for event in pg.event.get():
        # 被點擊
        if event.type == pg.MOUSEBUTTONUP:
            x, y = pg.mouse.get_pos()
            # 縮小40倍 0 40 80 120 -> 0 1 2 3
            x_scale, y_scale = x / x_inter, y / y_inter
            # 看到底左邊還又邊(0 1 2 3 4... 16)
            x_scale_round, y_scale_round = round(x_scale), round(y_scale)
            # 去除左右兩邊
            if (1 <= x_scale_round <= 15) and (1 <= y_scale_round <= 15):
                # 回到 40 80 120
                real_x, real_y = x_scale_round * x_inter, y_scale_round * y_inter
                # 應付寬高不一樣的情況, 取比較短的那個當作圓的半鏡
                circle_radius = min(x_inter / 2, y_inter / 2)
                if game_round % 2 == 0:
                    player_color = [0, 0, 0]
                else:
                    player_color = [255, 255, 255]
                pg.draw.circle(bg, player_color, [real_x, real_y], circle_radius)
                # 把某個圖層貼到上一層
                screen.blit(bg, [0, 0])
                # 對畫面進行更新(才會真的秀出來)
                pg.display.update()
                game_round = game_round + 1

        # 如果收到的事件是按x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False
pg.quit()
```
